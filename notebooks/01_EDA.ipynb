```python
# import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# load data
df = pd.read_csv("../data/raw/dataset.csv")
```

```python
# diplay info
print("====info====")
print(df.info())
# display ship
print("====shape====")
print(df.shape)
# display head
print("====head====")
display(df.head())
# dtypes
print("====dtypes====")
print(df.dtypes)
```

```python
# descriptive statistics
display(df.describe())
categorical_list = ['Gender', 'School_Type', 'Extracurricular_Activities', 'Internet_Access', 'Learning_Disabilities', 'Peer_Influence', 'Parental_Involvement', 'Access_to_Resources', 'Motivation_Level', 'Family_Income', 'Teacher_Quality', 'Parental_Education_Level', 'Distance_from_Home']
for category in categorical_list:
    print(f"==={category}====")
    print(df.value_counts(category))
```

## detect missing values and the duplicate

```python
# missing value

list_missing = ['Teacher_Quality', 'Parental_Education_Level', 'Distance_from_Home']

for miss in list_missing:
    print(f"==={miss} count===")
    print(df[miss].isna().sum())

# duplicates
print(f"===duplicated count===")
print(df.duplicated().sum())
```

## Histograms of Exam_Score and Hours_Studied

```python
# Exam_score

df["Exam_Score"].hist(bins=20)

plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")
plt.show()

# Hours_Studied
df["Hours_Studied"].hist(bins=20)

plt.xlabel("Hours Studied")
plt.ylabel("Number of Students")
plt.title("Distribution of Hours Studied")
plt.show()
```

## correlation heatmap before encoding

```python
numerical_df = df.select_dtypes(include="number")

correlation = numerical_df.corr()

plt.figure(figsize=(10, 8))

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```
